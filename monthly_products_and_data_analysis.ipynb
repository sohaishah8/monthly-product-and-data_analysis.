```python
import numpy as np
import pandas as pd
customers_data={
    'customer_id':['c001','c0002','c003','c004','c005'],
    'name':['ahmed','iqbal','sadeem','labeeb','shahmeer'],
    'city':['karachi','lahore','islmbad','karachi','lahore']
}
customers=pd.DataFrame(customers_data)
print(customers)
products_data={
    'product_id':['p001','p002','p003','p004','p005'],
    'product_name':['laptop','mouse','keyboard','T.V','LCD'],
    'category':['electronic','accessories','accessories','elcctronic','electronic'],
    'price':[30000,5000,6000,60000,32000]
}
products=pd.DataFrame(products_data)
print(products)
order_data={
    'order_id':['o001','o002','o003','o004','o005','o006','o007','o008','o009','o0010'],
    'customer_id':['c001','c002','c003','c004','c005','c006','c007','c008','c009','c0010'],
    'product_id':['p001','p002','p003','p004','p005','p006','p007','p008','p009','p0010'],
    'quantity':[40,566,788,67,87,90,789,90,657,5467],
    'order_date':[
        '2026-01-01',
        '2026-01-02',
        '2026-01-03',
        '2026-01-04',
        '2026-01-05',
        '2026-01-06',
        '2026-01-07',
        '2026-01-08',
        '2026-01-09',
        '2026-01-10'
    ]
}
orders=pd.DataFrame(order_data)
print(orders)
# creat csv file
customers.to_csv('customers.csv',index=False)
products.to_csv('products.csv',index=False)
orders.to_csv('orders.csv',index=False)
# read the file
customers=pd.read_csv('customers.csv')
products=pd.read_csv('products.csv')
orders=pd.read_csv('orders.csv')
#check karna
print(customers.head())
print(products.head())
print(orders.head())
# data inspection
print(customers.info())
print(products.info())
print(orders.info())
#missing value
print(customers.isnull().sum())
print(products.isnull().sum())
print(orders.isnull().sum())
#duplicated value
print(customers.duplicated().sum())
print(products.duplicated().sum())
print(orders.duplicated().sum())
#data time ma convert karna
orders['order_date']=pd.to_datetime(orders['order_date'])
#check
print(orders.dtypes)
#marge karna customer and order
df = orders.merge(
                 customers,
                 on ='customer_id',
                 how='left')
#merge karna ha product ko
df=df.merge(
         products,
          on='product_id',
          how='left'
    
)
print(df)
#calculate karna revenue($)
df['revenue']=(df['price']*df['quantity'])
#check krna
print(
      df[[ 'order_id',
           'product_name',
           'quantity',
            'revenue']])
#numpy se revenue ko classification krna
df['revenue_type']=np.where(
        df['revenue'] >= 500,
        'high',
        'Normal'
)
#check
print([[
        'product_name'
           'revenue'
        'revenue_type' ]])
#monthly sales analysis
df['month']= df['order_date'].dt.to_period('M')
#Monthly revenue:
monthly_sales = df.groupby('month')['revenue'].sum().reset_index()
print(monthly_sales)
#monthly_growth
monthly_sales['growth_%']=(monthly_sales['revenue'].pct_change().mul(100).round(2))
print(monthly_sales)
#top product
top_products = df.groupby('product_name').agg(
    total_quantity=('quantity','sum'),
    total_revenue=('revenue','sum')
).sort_values('total_revenue',ascending=False)
print(top_products)
#category_analysis
category_analysis=df.groupby('category').agg(
    total_sales=('revenue','sum'),
    units_sold=('quantity','sum')
).sort_values('total_sales',ascending=False)
print(category_analysis)
#customer analysis
customer_analysis=df.groupby(['customer_id','name']).agg(
    total_orders=('order_id','nunique'),   
    total_spent=('revenue','sum'),
    total_item=('quantity','sum')
).sort_values('total_spent',ascending=False)
print(customer_analysis)
#city analysis
city_analysis=df.groupby('city').agg(
    revenue=('revenue','sum'),
    customers=('customer_id','nunique')
).sort_values('revenue',ascending=False)
print(city_analysis)
#repeat customers
repeat_customers=customer_analysis[customer_analysis['total_orders']>1]
print(repeat_customers)
#final KAPI's
total_revenue=df['revenue'].sum()
total_orders=df['order_id'].nunique()
total_customers=df['customer_id'].nunique()
total_products=df['product_id'].nunique()
average_order_value=np.divide(total_revenue,total_orders)
#take print of variable
print(f'Total revenue:{total_revenue}')
print(f'Total_orders:{total_orders}')
print(f'Total_customers:{total_customers}')
print(f'Total_products:{total_products}')
print(f'Average orders:{round(average_order_value,2)}')

```
